**in vs range**

- if we want to access the elements directly use **in** operator

   for i in 'python' where i is characters

- if we want to access the elements using **range** apply **index**

      for in range(len('python')) where i is a number

```python
s1='python'
for i in s1:
    print(i,end=' ')
```

```python
s1='python'
for i in range(len(s1)):
    print(s1[i],end=' ')
```
